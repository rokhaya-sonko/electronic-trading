# Order Book Engine

We implement a stateful Level 2 order book.

Design principles:

- Event-driven updates
- Snapshot initialization
- Incremental update handling
- No DataFrame loops
- Real-time feature extraction


```python
class OrderBook:
    
    def __init__(self):
        self.bids = {}
        self.asks = {}
        self.prev_bid_size = None
        self.prev_ask_size = None
        
    def apply_snapshot(self, bids, asks):
        self.bids = {float(p): float(s) for p, s in bids}
        self.asks = {float(p): float(s) for p, s in asks}
        
    def apply_update(self, side, price, size):
        book = self.bids if side == "buy" else self.asks
        price = float(price)
        size = float(size)
        
        if size == 0:
            book.pop(price, None)
        else:
            book[price] = size
    
    def best_bid(self):
        return max(self.bids.keys())
    
    def best_ask(self):
        return min(self.asks.keys())
    
    def mid(self):
        return (self.best_bid() + self.best_ask()) / 2
    
    def spread(self):
        return self.best_ask() - self.best_bid()
    
    def imbalance(self):
        bid = self.best_bid()
        ask = self.best_ask()
        qb = self.bids[bid]
        qa = self.asks[ask]
        return (qb - qa)/(qb + qa)
    
    def microprice(self):
        bid = self.best_bid()
        ask = self.best_ask()
        qb = self.bids[bid]
        qa = self.asks[ask]
        return (ask*qb + bid*qa)/(qb + qa)
    
    def ofi(self):
        bid = self.best_bid()
        ask = self.best_ask()
        qb = self.bids[bid]
        qa = self.asks[ask]
        
        if self.prev_bid_size is None:
            self.prev_bid_size = qb
            self.prev_ask_size = qa
            return 0
        
        delta_bid = qb - self.prev_bid_size
        delta_ask = qa - self.prev_ask_size
        
        self.prev_bid_size = qb
        self.prev_ask_size = qa
        
        return delta_bid - delta_ask
    
    def depth_imbalance(self, levels=5):
        bid_prices = sorted(self.bids.keys(), reverse=True)[:levels]
        ask_prices = sorted(self.asks.keys())[:levels]
        
        qb = sum(self.bids[p] for p in bid_prices)
        qa = sum(self.asks[p] for p in ask_prices)
        
        return (qb - qa)/(qb + qa)
```

#### Live Data Acquisition

We connect to the Coinbase WebSocket Level 2 feed.

Workflow:
1. Receive snapshot.
2. Apply incremental updates.
3. Compute features at each update.
4. Store event-time data.

We collect data for a fixed time window (e.g., 5 minutes).

```python
book = OrderBook()
records = []

def on_message(ws, message):
    global book, records
    
    data = json.loads(message)
    
    if data['type'] == 'snapshot':
        book.apply_snapshot(data['bids'], data['asks'])
    
    elif data['type'] == 'l2update':
        for change in data['changes']:
            side, price, size = change
            book.apply_update(side, price, size)
        
        try:
            records.append([
                time.time(),
                book.mid(),
                book.spread(),
                book.imbalance(),
                book.microprice(),
                book.ofi(),
                book.depth_imbalance(5)
            ])
        except:
            pass

def on_open(ws):
    subscribe_message = {
        "type": "subscribe",
        "product_ids": ["BTC-USD"],
        "channels": ["level2"]
    }
    ws.send(json.dumps(subscribe_message))

ws = websocket.WebSocketApp(
    "wss://ws-feed.exchange.coinbase.com",
    on_message=on_message,
    on_open=on_open
)

thread = threading.Thread(target=ws.run_forever)
thread.start()

time.sleep(300)  # collect 5 minutes

ws.close()
thread.join()
```

##### Data Preparation

We transform event-time observations into a research dataset. We compute: ΔMid_t = Mid_{t+1} − Mid_t and remove missing values.

```python
df = pd.DataFrame(records, columns=[
    "timestamp","mid","spread",
    "imbalance","microprice",
    "ofi","depth_imbalance"
])

df["future_return"] = df["mid"].shift(-1) - df["mid"]
df = df.dropna()
```
